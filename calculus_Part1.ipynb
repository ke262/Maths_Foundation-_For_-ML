```python
import numpy as np
import matplotlib.pyplot as plt
```

1. A simple Line y = 2x+1

```python
def line_func(x):
    return 2*x+1
```

```python
x = np.linspace(-3, 3, 100)
y = line_func(x)
```

```python
print(x[:5])
print(y[:5])
```

```python
#given X and Y value, find the slope of the line
plt.figure(figsize=(6,4))
plt.plot(x,y)
plt.title("Line Plot")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(alpha=0.3)
plt.show()


# choose two points on the line
x1,x2 = -2, 1
y1,y2 = line_func(x1), line_func(x2)

rise = y2 - y1
run = x2 - x1

slope = rise/run

print("-"*50)
print("Line Example")
print("Point 1:", f"({x1},{y1})")
print("Point 2:", f"({x2},{y2})")
print("Rise:", rise)
print("Run:", run)
print("Slope (rise/run):", slope)
print("-"*50)
```

2. Curve y=x^2

```python
def curve(x):
    return x**2

x_vals = np.linspace(-3, 3, 200)
y_vals = curve(x_vals)

plt.figure(figsize=(6,4))
plt.plot(x_vals, y_vals)
plt.title("Curve: y^2")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(alpha=0.3)
plt.show()
```

3. Estimation slope with a small change in x

idea: if we move x a tiny bit and see how much y changes, that changes in y divided by change in x given an approximate slope.

slope = dy/dx

1 dy = change in y
2 dx = change in x

```python
def slope_estimate(func, x1, h=0.00001):
    x2 = x1+h
    y1 = curve(x1)
    y2 = func(x2)
    dy = y2-y1
    dx = h
    slope = dy/dx
    return slope
```

```python
points = [-2, 0, 2]

print("Slope estimation for y=x^2 using small change in x")

for p in points:
    s = slope_estimate(curve, p)
    print(f"x = {p:.1f} -> estimated slope = {s:.2f}")
```
