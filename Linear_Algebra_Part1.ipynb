```python
import numpy as np
```

1.Scalar,Vector,Matrix & Tensor in Numpy

```python
# scalar - single number
scalar = 5.0
print(scalar)
```

```python
# vector - 1D data
vector = np.array(
    [3, 7, 2]
)
print("Vector:",vector)
print("Shape of vector:", vector.shape)
```

```python
#matrix
matrix = np.array(
    [
        [1, 2, 3],
        [4, 5, 6]

    ]
    
)
print("Matrix:")
print(matrix)
print("Shape of matrix:", matrix.shape)
```

```python
# Tensor
tensor = np.array(
    [
        [
            [1, 2, 3],
            [4, 5, 6]
        ],
        [
            [11, 12, 13],
            [14, 15, 16]
        ]
    ]
)
print("Tensor:")
print(tensor)
print("Shape of tensor:", tensor.shape)
```

2.Creating vectors and matrices for ML style data

```python
X = np.array(
    [
        [2, 30],
        [5, 80],
        [1, 20]
    ]
)

y = np.array([0, 1, 0])

print("Data matrix X:\n", X)
print("X shape:", X.shape)

print("Output Vector y:\n", y)
print("y shape:", y.shape)

x1 = X[0]
print("\nFirst sample as a vector:", x1)
print("x1 shape:", x1.shape)
```

3.Basic operations: addition, scalar multiplication, transpose

```python
# Two vectors with same length
v1 = np.array([1, 2, 3])
v2 = np.array([4, 5, 6])

print("v1:", v1)
print("v2:", v2)

# vector addition
v_add = v1 + v2
print("\nVector addition v1 + v2:", v_add)

# scalar multiplication
v1_scaled = 2 * v1
print("Scalar multiplication 2 * v1:", v1_scaled)
```

```python
# matrix
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [10, 20],
    [30, 40]
])
print("\nMatrix A:\n", A)
print("Matrix B:\n", B)

#Matrix Addition
A_add_B = A+B
print("\nMatrix addition A+B:\n", A_add_B)

#Scaled multiplication
A_scaled = 0.5 * A
print("Scalar multiplication 0.5 * A:\n", A_scaled)

# transpose
A_T = A.T
print("\nTranspose of A:\n", A_T)
print("A Shape:", A.shape,"(A.T shape:", A_T.shape)
```
